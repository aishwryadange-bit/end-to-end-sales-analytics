```python
# data load
import pandas as pd
df = pd.read_excel(r"C:\Users\ASUS\OneDrive\Desktop\Jupyter projects\sales_raw_500.xlsx")
```

```python
df
```

```python
# Remove duplicates
df = df.drop_duplicates()
```

```python
df
```

```python
# Convert Order_Date to datetime
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
```

```python
# Remove rows where Order_Date is missing
df = df.dropna(subset=['Order_Date'])
```

```python
# Calculate Profit
df['Profit'] = df['Sales'] - df['Cost']
```

```python
# Extract Year and Month
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
```

```python
df
```

```python
df.to_csv(r"C:\Users\ASUS\OneDrive\Desktop\Jupyter projects\sales_cleaned.csv",index=False)
```
